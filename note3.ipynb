```python
# Q8

def larger(a,b,c):
  lrg = a
  if b>lrg:
    lrg = b
  if c > lrg:
    lrg = c

  return lrg
```

```python
p = int(input("enter p : "))
q = int(input("enter q : "))
r = int(input("enter r : "))

z = larger(p,q,r)
print(z)
```

```python
# Q10

def filterEven(lst):
  even = []
  for num in lst:
    if num%2 == 0:
      even.append(num)
  return even


list1 = [31,54,83,46,98,62]

ans = filterEven(list1)
print(ans)
```
