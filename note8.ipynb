A **lambda** is a small, unnamed (anonymous) function written in a single line.

```python
# lambda arguments: expression

sqr = lambda x: x **2

print(sqr(5))
```

```python
add = lambda x,y:x+y

add(3,4)
```

```python
marks = [45,78,32,90,55,28]

list(filter(lambda x:x>=40,marks))
```

```python
numbers = [4,7,13,16,72,84,56,93,27,28,94]

list(filter(lambda x:x%4==0,numbers))
```

##  A decorator is a function that wraps another function to add extra behavior — without touching the original function's code

```python
import time

def decorator(func):
  def wrapper():
    print("START")
    time.sleep(5)
    func()
    time.sleep(5)
    print("END")

  return wrapper
```

```python
@decorator
def greet():
  print("Hello")
```

```python
greet()
```

```python
def view_profile():
  print("name = john")
  print("post1 : post1 content")
  print("post2 : post2 content")
  print("post3 : post3 content")
```

```python
def follow_required(func):
  def wrapper():
    if is_follow:
      func()
    else:
      print("Follow first")
  return wrapper
```
