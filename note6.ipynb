## json

```python
import json
```

```python
student = {
    "id": 101,
    "name": "John",
    "age": 20,
    "course": "Computer Science",
    "grade": "A",
    "city": "Kochi"
}


data = json.dumps(student)
print(student)
print(data)

print(type(student))
print(type(data))

file = open("student_data.json","a")
file.write(data)
file.close()


#with open("student_data.json","a") as file:
#    file.write(data)
```

```python
student = {
    "id": 101,
    "name": "John",
    "age": 20,
    "course": "Computer Science",
    "grade": "A",
    "city": "Kochi"
}


data = json.dumps(student)

print(type(student))
print(type(data))


load_data = json.loads(data)
print(load_data)
print(type(load_data))
```
