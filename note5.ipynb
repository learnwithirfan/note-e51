```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def __str__(self):
    return f"{self.name}({self.age})"

  def greet(self):
    print("Hello, my name is " + self.name)
```

```python
class Circle:
    def __init__(self,radius):
      self.radius = radius

    @property
    def diameter(self):
      return 2*self.radius

    @property
    def area(self):
      return 3.14*self.radius**2
```

```python
c1 = Circle(3)

print(c1.diameter)
print(c1.area)
```

```python
class Staff:
  def __init__(self, name, salary):
    self.name = name
    self.__salary = salary

  def __str__(self):
    return self.name

  def get_salary(self):
    return self.__salary

  def set_salary(self, salary):
    self.__salary = salary
```

```python
s1 = Staff("alice",48000)
s2 = Staff("bob",45000)

print(s1.__salary)
```

```python
print(s1.get_salary())
```

```python
s1.set_salary(50000)

print(s1.get_salary())
```

```python
class Staff:

  count = 0

  def __init__(self, name, salary):
    self.name = name
    self.__salary = salary
    Staff.count += 1

  @classmethod
  def get_count(cls):
    return cls.count


  def __str__(self):
    return self.name

  def get_salary(self):
    return self.__salary

  def set_salary(self, salary):
    self.__salary = salary

  @staticmethod
  def department():
    print("IT department")
```

```python
Staff.department()
```

```python
s1 = Staff("alice",48000)
s2 = Staff("bob",45000)
s3 = Staff("Roy",63000)


Staff.get_count()
```

```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def __str__(self):
    return f"{self.name}({self.age})"

  def greet(self):
    print("Hello, my name is " + self.name)
```

```python
class Student(Person):
  def __init__(self, name, age, student_id):
    super().__init__(name, age)
    self.student_id = student_id

  def __str__(self):
    return f"{super().__str__()},{self.student_id}"
```
