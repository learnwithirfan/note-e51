## string slice

```python
s = "python"

print(s[3])
print(s[-2])
print(s[2:5])
# [start:end:step]
print(s[0:6:2]) # pto

print(s[:]) # inter
print(s[:4])
print(s[::3])

print(s[::-1]) # interv | dsa | reverse
```

```python
l2 = ['apple','banana','orange']
print(l2)
l2.append("grapes")
print(l2)
l2.insert(1,'lemon') # insert(index,value)
print(l2)
l2.remove("banana")
print(l2) # ['apple', 'lemon', 'orange', 'grapes']
a = l2.pop() # delete last value
print(a) # grapes
print(l2) # ['apple', 'lemon', 'orange']

print(l2.index('apple'))

l3 = ['apple','apple','grapes']
print(l3.count('apple'))

l4 = [426,735,243,82]
l4.sort()
print(l4)

l5 = [735,243,82]
l5.reverse()
print(l5)
```
