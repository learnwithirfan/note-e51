## set()

```python
s1 = set()
s1.add(3)
s1.add(4)
s1.add(6)
s1.add(1)
s1.add(4)
s1.add(6)

print(s1)
print(type(s1))
```

```python
s2 = set([1,2,4,5,7,4,2,1])
print(s2)
```

```python
s3 = set()
for i in range(5):
  x = int(input("enter your number : "))
  s3.add(x)

print(s3)
```

```python
s5 = set([1,4,7])
s6 = set([4,9,2])

print(s5 | s6)
print(s5 & s6)
print(s5.union(s6))
print(s5 - s6)
```

```python
def greet():
  print("Hello , Welcome to Oooty")
  print("How are you")

greet()
# no param , no return
```

```python
def greet(name):
  print(f"Hello {name}.Welcome to Ooty")

greet("alice")
```

```python
def greet(name='user'):
  print(f"Hello {name}.Welcome to Ooty")

greet("alice")
greet()
# params-ok ,no-return
```

```python
def area_circle(radius):
  area = 3.14 * radius * radius
  return area

a1 = area_circle(4)
a2 = area_circle(6)


print(a1)
print(a2)
```

```python
rad = int(input("enter radius : "))
area = area_circle(rad)
print(area)
```
